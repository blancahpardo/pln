## 🔹 1. Importación de librerías y carga de texto

```python
import re

texto = """
The patient is actually very sensible to the drug. El paciente es actualmente muy sensible al medicamento. 
Eventually, the physician decided to assist the patient. Eventualmente, el médico decidió asistir al paciente.
The current treatment has no effect. El tratamiento actual no tiene efecto.
This is an actual problem that needs attention. Este es un problema actual que requiere atención.
Eventually, the patient decided to look for a second opinion. Al final, el paciente decidió buscar una segunda opinión.
"""
print("Texto de entrada:\n", texto)
```

## 🔹 2. Detección de un posible falso amigo específico

```python
patron = r"\bactual\b"
coincidencias = re.findall(patron, texto, flags=re.IGNORECASE)
print(f"Número de ocurrencias de 'actual': {len(coincidencias)}")
```

```python
oraciones = texto.split('\n')

for oracion in oraciones:
    oracion = oracion.strip()
    if oracion:
        if re.search(patron, oracion, flags=re.IGNORECASE):
            oracion_resaltada = re.sub(patron, "actual", oracion, flags=re.IGNORECASE)
            print(f"\n📝 Oración con 'actual': {oracion_resaltada}")
```

## 🔹 3. Detección automática de posibles falsos amigos a partir de un glosario correcto

### Búsqueda de ocurrencias (nº) de posibles falsos amigos en el texto (a partir de glosario de falsos amigos)

```python
falsos_amigos = {
    "actual": "actual",
    "actually": "actualmente",
    "sensible": "sensible",
    "eventually": "eventualmente",
    "assist": "asistir",
    "library": "librería"
}
```

```python
for palabra in falsos_amigos:
    print(f"Buscando: '{palabra}'")
    resultados = re.findall(fr"\b{palabra}\b", texto, flags=re.IGNORECASE)
    print(f"  > Ocurrencias: {len(resultados)}\n")
```

### Búsqueda de posibles falsos amigos por cada oración (a partir de glosario de falsos amigos)

```python
print("\n=== DETECCIÓN DE POSIBLES FALSOS AMIGOS POR ORACIONES ===")

oraciones = texto.split('\n') 
for oracion in oraciones: 
    if oracion.strip(): 
        print(f"\n📝 Oración: {oracion}")
        for palabra in falsos_amigos: 
            if re.search(fr"\b{palabra}\b", oracion, flags=re.IGNORECASE):
                print(f"   ⚠️ Término en inglés susceptible de falso amigo: '{palabra}' → Falso amigo en español: {falsos_amigos[palabra]}")
```

### Búsqueda de posibles falsos amigos traducidos por cada oración + propuesta de traducción correcta

```python
falsos_amigos_corr = {
    "actual": "real",
    "actually": "en realidad, realmente...",
    "sensible": "sensato",
    "eventually": "al final, finalmente, en última instancia...",
    "assist": "ayudar",
    "library": "biblioteca"
}
```

```python
print("\n=== DETECCIÓN DE POSIBLES FALSOS AMIGOS POR ORACIONES + PROPUESTA DE EQUIVALENTE CORRECTO ===")

oraciones = texto.split('\n') 
for oracion in oraciones:
    if oracion.strip():
        print(f"\n📝 Oración: {oracion}")
        for palabra in falsos_amigos:
            if re.search(fr"\b{palabra}\b", oracion, flags=re.IGNORECASE): #*
                print(f"   ⚠️ Contiene posible falso amigo: '{palabra}' → traducción correcta: {falsos_amigos_corr[palabra]}")
```

## 🔹 4. Detección de traducciones incorrectas por falso amigo + propuesta de traducción correcta

```python
import re

print("\n=== ORACIONES CON FALSOS AMIGOS REALES DETECTADOS (TO + TM) Y PROPUESTA DE EQUIVALENCIA CORRECTA ===")

oraciones = texto.split('\n')

for oracion in oraciones:
    if oracion.strip():  
        for palabra in falsos_amigos:
            traduccion_erronea = falsos_amigos[palabra]
            if (re.search(fr"\b{palabra}\b", oracion, flags=re.IGNORECASE) and 
                re.search(fr"\b{traduccion_erronea}\b", oracion, flags=re.IGNORECASE)):

                print(f"\n📝 Oración: {oracion}")
                print(f"   ⚠️ Falso amigo detectado: '{palabra}' ↔ '{traduccion_erronea}'")
                print(f"   ✅ Traducción correcta sugerida: {falsos_amigos_corr[palabra]}")
```
