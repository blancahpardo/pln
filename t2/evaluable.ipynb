### ✍️ Actividad 1

1. Crea con un asistente de IA generativa 5-10 frases en inglés que contengan varias de las palabras de tu diccionario de falsos amigos. A continuación, tradúcelas al español asegurándote de que se cometen falsos amigos. Coloca cada oración en inglés seguida de su traducción al español (del mismo modo que hicimos con los textos de práctica en el cuaderno principal de trabajo). 
Condición: Una de las oraciones debe tener su traducción correspondiente en español sin ningún falso amigo, es decir, la traducción de una de las oraciones debe ser correcta.

2. Haz una lista de falsos amigos (puedes tomar la que generamos en el cuaderno principal de trabajo de la presente unidad), añadiendo algunos más que se te ocurran en función de tu experiencia o investigación.

3. Ejecuta un código que identifique de todo el texto las oraciones EN-ES que contengan una de las palabras en inglés de tu diccionario de falsos amigos. Tiene que mostrar la oración u oraciones en las que aparece, además de un recuento del total de ocurrencias de dicha palabra.

4. Ejecuta un código que identifique el número de veces que aparecen las palabras en inglés de tu diccionario de falsos amigos.

5. Ejecuta un código que identifique las oraciones EN-ES que contengan posibles falsos amigos (palabras en inglés) según tu diccionario de falsos amigos.

### ✍️ Actividad 2:
