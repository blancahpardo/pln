# 📘 Introducción a Regex (Expresiones Regulares) en Python

**📌 ¿Qué es Regex?**

Las expresiones regulares (o regex) son una herramienta muy poderosa que sirve para buscar, extraer o reemplazar texto de forma flexible. Son muy útiles en el Procesamiento del Lenguaje Natural (PLN) cuando queremos:
- Buscar palabras o frases específicas en un texto.
- Limpiar datos (quitar puntuación, espacios extra, etc.)
- Identificar patrones (como fechas, correos electrónicos, etc.)

No se busca una palabra exacta, sino un patrón de forma general. Son como una minilengua con reglas para identificar cosas en el texto.

Las expresiones regulares te ayudarán a buscar patrones lingüísticos en grandes corpus de texto. Además, son especialmente útiles para limpiar textos, buscar estructuras o crear filtros de información.

Aunque al principio pueden parecer extrañas, ¡con la práctica se vuelven una herramienta indispensable!

**📌 ¿Por qué son útiles para la programación y el PLN?**

En Procesamiento del Lenguaje Natural (PLN), trabajamos con mucho texto. Regex sirve para:

* Limpiar textos antes de analizarlos.

* Detectar palabras que tienen una forma concreta (verbos, nombres propios, números…).

* Extraer información útil de textos grandes: correos, fechas, cifras, palabras clave…

**📌 ¿Y por qué son importantes para un futuro lingüista computacional?**

Un lingüista computacional o un lingüista que trabaja con traducción automática, corpus lingüísticos, corrección ortográfica, chatbots, búsqueda de información… necesita extraer patrones del lenguaje de forma rápida y precisa. Y eso lo hace con ayuda de herramientas como Regex.

Por ejemplo:

* Crear una herramienta que busque verbos en infinitivo dentro de un corpus.

* Analizar estilos lingüísticos en traducciones.

* Encontrar errores frecuentes en textos escritos.

Aunque ahora lo estamos viendo con ejemplos simples, estos conceptos se usan en muchas herramientas profesionales, como motores de traducción automática, sistemas de corrección, minería de textos, etc.

```python
import re 
```

```python
texto = "Mi gato duerme todo el día."

resultado = re.search("gato", texto)

print(resultado)
```

**🧠 ¿Qué ha pasado aquí?**

`re.search()` busca la primera aparición del patrón (en este caso "gato") dentro del texto.

Si lo encuentra, devuelve un objeto especial `(Match)`; si no, devuelve `None`.



# 📋 Empecemos a trabajar con Regex

```python
if resultado:
    print("¡Encontré la palabra 'gato'!")
else:
    print("No encontré la palabra 'gato'.")
```

## 🎯 Búsqueda de palabras que empiezan por una letra concreta

```python
texto2 = "El coche y la casa están cerca del campo."

palabras_c = re.findall(r"\bc\w+", texto2)

print("Palabras que empiezan por 'c':", palabras_c)
```

## 🔄 Reemplazar palabras con Regex

```python
texto3 = "María tiene un perro. El perro se llama Max."

nuevo_texto = re.sub("perro", "gato", texto3)

print(nuevo_texto)
```

## 🧹 Limpiar puntuación

```python
texto4 = "¡Hola! ¿Cómo estás? Estoy bien, gracias."

texto_limpio = re.sub(r"[^\w\s]", "", texto4)

print("Texto sin puntuación:", texto_limpio)
```

## 🔍 Buscar todas las apariciones de una palabra

```python
texto5 = "El gato está en el tejado. El gato duerme."

apariciones = re.findall("gato", texto5)

print("Número de veces que aparece 'gato':", len(apariciones))
print("Lista de coincidencias:", apariciones)
```

## 🧪 Buscar correos electrónicos

```python
texto6 = "Puedes escribir a juan@example.com o a maria@gmail.com."

correos = re.findall(r"\w+@\w+\.com", texto6)

print("Correos encontrados:", correos)
```

## ✅ Encontrar todas las palabras que terminan en "-ar" (como en los verbos en infinitivo)

```python
texto = "Cantar, bailar, comer y vivir son verbos comunes."

palabras_ar = re.findall(r"\b\w+ar\b", texto)

print("Palabras que terminan en 'ar':", palabras_ar)
```

## ✅ Contar cuántas palabras hay en una frase

```python
frase = "La lingüística computacional es fascinante."

palabras = re.findall(r"\b\w+\b", frase)

print("Palabras encontradas:", palabras)
print("Número total de palabras:", len(palabras))
```

## ✅ Detectar mayúsculas iniciales (nombres propios, principios de oración)

```python
texto = "Madrid es la capital de España. Juan estudia Lenguas Modernas."

mayusculas = re.findall(r"\b[A-Z]\w+", texto)

print("Palabras con inicial mayúscula:", mayusculas)
```

## ✅ Verificar si una palabra aparece al principio de una frase

```python
frase = "Hola, ¿cómo estás?"

if re.match(r"^Hola", frase):
    print("La frase empieza por 'Hola'")
else:
    print("La frase NO empieza por 'Hola'")
```

## ✅ Extraer números en un texto

```python
texto = "Tengo 2 gatos, 1 perro y 3 peces."

numeros = re.findall(r"\d+", texto)

print("Números encontrados:", numeros)
```

```python
texto = "Tengo 2 gatos, 1 perro y 3 peces. Mi teléfono es el 679483750."

numeros = re.findall(r"\d{4,}", texto)

print("Números encontrados:", numeros)
```

```python
texto = "Tengo 2 gatos, 1 perro y 3 peces. Mi teléfono es el 679483750 y mi pin 1234."

numeros_prueba1 = re.findall(r"\d{4}", texto)

print("Números encontrados en la prueba 1:", numeros_prueba1)
```

```python
numeros_prueba2 = re.findall(r"\b\d{4}\b", texto)

print("Números encontrados en la prueba 2:", numeros_prueba2)
```

```python
texto = "Tengo 2 gatos, 1 perro y 3 peces. Mi teléfono es el 679483750 y mi pin 1234."

texto_sin_numeros = re.sub(r"\d+", "", texto)

print("Texto sin números:", texto_sin_numeros)
```

## 📌 Resumen para recordar

| Expresión | Significado                                 |
| --------- | ------------------------------------------- |
| `.`       | Cualquier carácter (excepto salto de línea) |
| `\w`      | Letras, números o guion bajo                |
| `\s`      | Espacio en blanco                           |
| `\d`      | Dígito (0-9)                                |
| `+`       | Uno o más                                   |
| `*`       | Cero o más                                  |
| `\b`      | Límite de palabra                           |
| `[]`      | Conjunto de caracteres                      |
| `^`       | Comienzo de línea o negación en corchetes   |
| `$`       | Final de línea                              |
