## **🧾 Actividad propuesta**


#### Instrucciones

1. Con ayuda de un asistente de LLM (ChatGPT, Deepseek, Gemini...), crea 3 textos de temática institucional y similar, de entre 300 y 400 palabras que contengan:
  - Varias veces en el texto la palabra "resiliencia".
  - Varios dobles espacios.
  - Varias ausencias incorrectas de espacio antes de coma o punto.
  - Varias estructuras "sustantivo + preposición 'a' + verbo en infinitivo"
  - Varios gerundios con uso de posterioridad y/o con valor de adjetivo (es decir, gerundios con usos no aceptados por la RAE).
  - En el texto 1 se debe emplear una o varias veces la palabra 'adquirir'; en el texto 2 se debe emplear una o varias veces la palabra 'obtener'; en el texto 3 se debe emplear una o varias veces la palabra 'conseguir'.
  - En el texto 1 se debe emplear una o varias veces la palabra 'trabajador'; en el texto 2 se debe emplear una o varias veces la palabra 'empleado'; en el texto 3 se debe emplear una o varias veces la palabra 'asalariado'.

Es importante que vayas generando un buen prompt para que el asistente cree textos coherentes y gramaticalmente correctos en español, pero con las solicitudes de contenido indicadas anteriormente. Si es necesario, deberás explicarle o ponerle ejemplos para que entienda lo que necesitas (por ejemplo, quizá tengas que investigar y explicarle lo que es un gerundio con uso de posterioridad, agramatical según la RAE).

Además, deberás explicarle al asistente que no quieres que aplique cada norma anterior una sola vez por cada código, sino de manera anárquica (más veces en un texto que en otro).

2. Una vez hayas obtenido los textos, deberás copiarlos y pegarlos como texto sin formato (.txt), y nombrarlos con la nomenclatura: discurso1.txt; discurso2.txt; discurso3.txt.

3. Comienza el ejercicio buscando esos ficheros .txt e imprimiéndolos por pantalla, como hicimos en el aula. Puedes emplear el entorno de Google Colab o bien en local (con VSC o similar). En clase vimos las dos formas de ejecutar código de ficheros que estén en nuestro ordenador.

4. Realiza con el código aprendido en este tema un análisis automático en los 3 textos para averiguar:
  - Frecuencia de aparición de la palabra "resiliencia".
  - Frecuencia y lugar de aparición de los dobles espacios.
  - Frecuencia y lugar de aparición de ausencia indebida de espacios (antes de coma o punto).
  - Casos de estructura "a + infinitivo".
  - Casos de gerundio.
  - Comparación entre el uso de los tripletes de palabras (adquirir vs. obtener vs. conseguir / trabajador vs. empleado vs. asalariado).

Nota: las variables de esta actividad ya no serán 'text', sino 'discurso', ni 'texto1', 'texto2', sino que deberán nombrarse 'discurso1', 'discurso2', 'discurso3', etc.

Puedes emplear el código más sencillo (para 3 textos, en lugar de las alternativas que también hemos enseñado para un número ilimitado de textos).

Ve trabajando todos los códigos en orden en las celdas de código siguientes según estas instrucciones.

Consejo: cuando elabores la tabla comparativa de términos entre discursos, te recomiendo que cambies el tabulador de '8' puntos a '10' puntos, y los guiones de '50' a '60'.

Tarea opcional de la actividad: Elabora automáticamente un informe final con los resultados de todas las acciones solicitadas gracias al uso de funciones definidas.
