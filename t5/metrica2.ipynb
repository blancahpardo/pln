## 1️⃣ ¿Qué es METEOR?

METEOR (*Metric for Evaluation of Translation with Explicit ORdering*) es una métrica que evalúa traducciones automáticas comparándolas con una traducción humana de referencia.

METEOR considera:
- Coincidencias exactas  
- Coincidencias por raíz  
- Sinónimos (solo en inglés)  
- Penalización por desorden  


⚠️ **Nota sobre el español:**  
METEOR funciona mejor en inglés porque usa WordNet. En español funciona, pero sin detección de sinónimos.

```python
import nltk
from nltk.translate.meteor_score import meteor_score

nltk.download('wordnet')
nltk.download('omw-1.4')
```

## 3️⃣ Ejemplo sencillo en español


```python
ref = "El gato duerme en la alfombra."
hyp1 = "El gato está durmiendo en la alfombra."
hyp2 = "El perro corre en el parque."

ref_tokens = ref.split()
h1_tokens = hyp1.split()
h2_tokens = hyp2.split()

meteor_hyp1 = meteor_score([ref_tokens], h1_tokens)
meteor_hyp2 = meteor_score([ref_tokens], h2_tokens)

print("Referencia:", ref_tokens)
print("Hipótesis 1:", h1_tokens)
print("Hipótesis 2 :", h2_tokens)
print("-" * 40)
print(f"METEOR (hipótesis 1): {meteor_hyp1:.3f}")
print(f"METEOR (hipótesis 2) : {meteor_hyp2:.3f}")
```

## 4️⃣ Función para evaluar METEOR


```python
def evaluar_meteor(ref_text, hyp_text):
    ref_tokens = ref_text.split()
    hyp_tokens = hyp_text.split()
    score = meteor_score([ref_tokens], hyp_tokens)
    print("Referencia:", ref_text)
    print("Hipótesis :", hyp_text)
    print("METEOR:", round(score,3))
```

```python
evaluar_meteor("El gato duerme en la alfombra.", "El gato está durmiendo en la alfombra.")
```

## 5️⃣ Ejemplo avanzado en contexto médico


```python
ref_med = "El médico recetó un nuevo tratamiento para controlar la condición crónica del paciente."
hyp_med_A = "El doctor recetó un nuevo tratamiento para controlar la enfermedad crónica del paciente."
hyp_med_B = "El hospital dio una medicina para resolver el problema del enfermo."

evaluar_meteor(ref_med, hyp_med_A)
print()
evaluar_meteor(ref_med, hyp_med_B)
```
