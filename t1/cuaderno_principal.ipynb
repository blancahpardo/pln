## 🔹 2. Variables y tipos de datos

```python
# Declaramos un texto:
texto1 = """El lenguaje claro es una herramienta fundamental para garantizar la accesibilidad informativa;
todos los ciudadanos tienen derecho a comprender los mensajes institucionales sin barreras lingüísticas."""
```

```python
# Forma 1 de mostrar en pantalla:
print("Texto de entrada:")
print(texto1)
```

```python
# Forma 2 de mostrar en pantalla:
print("Texto de entrada:", texto1)
```

## 🔹 3. Contar número de palabras

```python
palabras = texto1.split()
num_palabras = len(palabras)
print("\nNúmero de palabras:", num_palabras)
```

## 🔹 4. Contar signos de puntuación

```python
import string
signos = [c for c in texto1 if c in string.punctuation]
print("Signos de puntuación:", signos)
num_signos = len(signos)
print("Total de signos de puntuación:", num_signos)
```

## 🔹 5. Contar el número de oraciones

```python
oraciones = texto1.split(".")
num_oraciones = len(oraciones)
print("Número estimado de oraciones:", num_oraciones)
```

## 🔹 6. Calcular el número de sílabas

```python
import re
silabas = re.findall(r'[aeiouáéíóúü]+', texto1, re.IGNORECASE)
num_silabas = len(silabas)
print("Número aproximado de sílabas:", num_silabas)
```

## 🔹 7. Índice de legibilidad aproximado

```python
# Fórmula adaptada: Promedio de sílabas por palabra + promedio de palabras por oración
indice = (num_silabas / num_palabras) + (num_palabras / num_oraciones)
print(indice)
print("\nÍndice aproximado de legibilidad:", round(indice, 2))
```
