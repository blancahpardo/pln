## 1️⃣ ¿Qué es un término y por qué es difícil?

Un **término** es una unidad léxica (una palabra o expresión) que designa un concepto de un **dominio especializado**.

Ejemplos (dominio: PLN):
- *etiquetado morfosintáctico*
- *modelo de lenguaje*
- *tokenización*

### ¿Por qué la extracción automática no es trivial?
Porque hay “ruido”:
- Palabras frecuentes que **no** son términos: *de, la, y, tiene…*
- Palabras frecuentes en general que aparecen también en el dominio: *sistema, análisis…*
- Expresiones multi-palabra que sí son términos: *red neuronal*, *aprendizaje supervisado*

### Estrategia típica en PLN (interpretables)
1. Generar **candidatos** (tokens y n-gramas)
2. Filtrar (frecuencia mínima, stopwords, longitud, etc.)
3. Rankear (¿qué es más “propio” del dominio?)
4. Validar (humano / experto / tarea downstream)


## 2️⃣ Frecuencia absoluta y relativa, y comparación entre corpus

### Frecuencia absoluta
Cuántas veces aparece una unidad:
- `freq("tokenización") = 25`

### Frecuencia relativa
Frecuencia ajustada por el tamaño del corpus (muy importante):
- `freq_rel = freq / total_tokens`
- Por ejemplo, “tokenización” puede aparecer 25 veces en un corpus de 10.000 tokens (0.25%)
  y 5 veces en otro de 100.000 tokens (0.005%).  
  **Comparar absolutos sin normalizar engaña.**

### Comparación corpus general vs. especializado
Buscamos unidades que:
- aparezcan con **frecuencia relativamente alta en el corpus especializado**
- y/o estén **mucho menos presentes en el corpus general**

Esto da lugar a medidas de “termhood” (carácter terminológico).
En este cuaderno implementaremos una medida simple y muy usada en docencia:

**Weirdness (raridad relativa):**
\[
weirdness(t) = \frac{f_{esp}(t)/N_{esp}}{f_{gen}(t)/N_{gen} + \epsilon}
\]
- Si es grande ⇒ el término es mucho más típico del especializado.
- `ε` evita dividir entre 0.

### 3.3. Tokenización y lematización 

Aquí haremos una tokenización **simple y transparente**:
- pasamos a minúsculas
- quitamos URLs
- eliminamos signos/puntuación/números
- dividimos por espacios

> 🧠 No es “perfecto”, pero es ideal para entender el proceso.

```python
def simple_tokenize(text: str) -> List[str]:
    """
    Tokenizador simple:
    - pasa a minúsculas
    - elimina URLs
    - mantiene letras con tildes/ñ (rango Unicode básico latino)
    - convierte todo lo demás en espacios
    - divide por espacios
    - filtra tokens muy cortos (opcional)
    
    Nota didáctica:
    Esto NO es perfecto, pero es transparente y suficiente para practicar
    extracción terminológica basada en frecuencias.
    """
    text = text.lower()
    text = re.sub(r"https?://\S+|www\.\S+", " ", text)  # quita URLs
    
    # Reemplazamos por espacio cualquier cosa que NO sea letra (incluye tildes) o espacio
    text = re.sub(r"[^a-záéíóúüñ\s]", " ", text)
    
    # Colapsa espacios múltiples
    text = re.sub(r"\s+", " ", text).strip()
    
    tokens = text.split(" ")
    
    # Filtramos tokens muy cortos (p.ej. "a", "y" se filtran después con stopwords)
    tokens = [t for t in tokens if len(t) >= 2]
    return tokens


# Probamos el tokenizador con un ejemplo pequeño
example = "La tokenización en PLN: modelos, embeddings y n-gramas (2025)."
simple_tokenize(example)
```

### 3.4. Stopwords

Las **stopwords** son palabras muy frecuentes con poco contenido terminológico:
- artículos, preposiciones, conjunciones…

En extracción de términos, si no filtramos stopwords, el ranking se llena de:
- *de, la, en, y, que, los…*

Aquí usaremos una lista básica en español (editable).

> 📌 OJO: en terminología real, a veces interesa conservar “de” dentro de expresiones
> (véase la ampliación al final).


```python
# Lista básica (puedes ampliarla según el dominio)
STOPWORDS_ES = {
    "de","la","que","el","en","y","a","los","del","se","las","por","un","para",
    "con","no","una","su","al","lo","como","más","pero","sus","le","ya","o",
    "este","sí","porque","esta","entre","cuando","muy","sin","sobre","también",
    "me","hasta","hay","donde","quien","desde","todo","nos","durante","todos",
    "uno","les","ni","contra","otros","ese","eso","ante","ellos","e","esto",
    "mí","antes","algunos","qué","unos","yo","otro","otras","otra","él","tanto",
    "esa","estos","mucho","quienes","nada","muchos","cual","poco","ella","estar",
    "estas","algunas","algo","nosotros","mi","mis","tú","te","ti","tu","tus"
}

def remove_stopwords(tokens: List[str], stopwords=STOPWORDS_ES) -> List[str]:
    return [t for t in tokens if t not in stopwords]

# Ejemplo rápido
remove_stopwords(simple_tokenize("La tokenización es un paso fundamental en PLN."))
```

### 3.5. Preprocesado del corpus

```python
def preprocess_corpus(corpus: List[str]) -> List[str]:
    """
    Convierte una lista de textos en una lista plana de tokens:
    - tokeniza
    - elimina stopwords
    """
    all_tokens = []
    for doc in corpus:
        toks = simple_tokenize(doc)
        toks = remove_stopwords(toks)
        all_tokens.extend(toks)
    return all_tokens

tokens_esp = preprocess_corpus(corpus_especializado)
tokens_gen = preprocess_corpus(corpus_general)

len(tokens_esp), len(tokens_gen), tokens_esp[:20]
```

### 3.6. Extracción de n-gramas

Muchos términos importantes son **expresiones**:
- *modelo de lenguaje*
- *traducción automática*
- *red neuronal*

Un **n-grama** es una secuencia de `n` tokens consecutivos:
- unigramas: `n=1` → “tokenización”
- bigramas: `n=2` → “modelo lenguaje”
- trigramas: `n=3` → “traducción automática neuronal”

> Nota: al quitar stopwords, “modelo de lenguaje” se convierte en “modelo lenguaje”.

> Esto es normal en pipelines simples. En versión ampliada veremos cómo conservar “de”.


```python
def get_ngrams(tokens: List[str], n: int) -> List[Tuple[str, ...]]:
    """
    Devuelve una lista de n-gramas como tuplas.
    Ejemplo: tokens = ["modelo","lenguaje","grande"], n=2
    -> [("modelo","lenguaje"), ("lenguaje","grande")]
    """
    return [tuple(tokens[i:i+n]) for i in range(len(tokens)-n+1)]

# Ejemplo
get_ngrams(["modelo","lenguaje","grande"], 2)
```

### 3.7. Recuento frecuencia n-gramas

```python
def count_ngrams(tokens: List[str], n: int) -> Counter:
    """
    Cuenta frecuencias de n-gramas y devuelve Counter:
    Counter({('modelo','lenguaje'): 3, ...})
    """
    ngrams = get_ngrams(tokens, n)
    return Counter(ngrams)

freq1_esp = count_ngrams(tokens_esp, 1)
freq2_esp = count_ngrams(tokens_esp, 2)
freq3_esp = count_ngrams(tokens_esp, 3)

freq1_gen = count_ngrams(tokens_gen, 1)
freq2_gen = count_ngrams(tokens_gen, 2)
freq3_gen = count_ngrams(tokens_gen, 3)

freq2_esp.most_common(10)
```

### 3.8. Filtrado de candidatos y conversión a texto

Antes de rankear, filtramos “ruido”. Reglas típicas:
- **frecuencia mínima**: si aparece 1 vez, suele ser ruido (depende del tamaño del corpus)
- **longitud mínima**: evitar tokens demasiado cortos
- **caracteres raros**: ya los eliminamos en limpieza
- (opcional) **listas negras**: palabras muy generales del dominio (*sistema, análisis…*) si molestan

En este cuaderno aplicaremos filtros simples:
- unigramas: frecuencia ≥ 2
- bigramas/trigramas: frecuencia ≥ 2 (en corpus reales quizá ≥ 3 o ≥ 5)


```python
def ngram_to_text(ng: Tuple[str, ...]) -> str:
    """Convierte ('modelo','lenguaje') -> 'modelo lenguaje'"""
    return " ".join(ng)

def filter_by_min_freq(counter: Counter, min_freq: int) -> Dict[str, int]:
    """
    Filtra un Counter de n-gramas por frecuencia mínima y devuelve dict {texto: freq}.
    """
    out = {}
    for ng, f in counter.items():
        if f >= min_freq:
            out[ngram_to_text(ng)] = f
    return out

cand1_esp = filter_by_min_freq(freq1_esp, min_freq=2)
cand2_esp = filter_by_min_freq(freq2_esp, min_freq=2)
cand3_esp = filter_by_min_freq(freq3_esp, min_freq=2)

len(cand1_esp), len(cand2_esp), len(cand3_esp)
```

### 3.9. Frecuencia relativa

**Frecuencia relativa: normalizar por tamaño del corpus**

Si un corpus tiene más tokens, sus frecuencias absolutas serán mayores aunque el fenómeno sea el mismo.

Definimos:
- `N_esp = total tokens (tamaño) del corpus especializado`
- `N_gen = total tokens (tamaño) del corpus general`

Y la frecuencia relativa:
- `rel = freq / N`

Esto permite comparar.


```python
N_esp = len(tokens_esp)
N_gen = len(tokens_gen)

def rel_freq(freq: int, N: int) -> float:
    return freq / N if N > 0 else 0.0

# ejemplo
term = "tokenización"
freq_term = cand1_esp.get(term, 0)
rel_freq(freq_term, N_esp), freq_term
```

### 3.10. Raking por *weirdness*

**Ranking por “Weirdness” (comparación especializado vs. general)**:

*Weirdness* aumenta si:
- el término es relativamente frecuente en el corpus especializado
- y relativamente raro en el corpus general

Usamos:
\[
weirdness(t) = \frac{f_{esp}(t)/N_{esp}}{f_{gen}(t)/N_{gen} + \epsilon}
\]

**Advertencia didáctica:**
- Si el término no aparece en el corpus general, el denominador sería 0.  
  Por eso añadimos `ε` (un número muy pequeño).
- En corpora muy pequeños (como nuestro ejemplo), los rankings pueden ser inestables.
  En corpora reales mejora mucho.


```python
def compute_weirdness(
    cand_esp: Dict[str, int],
    freq_gen_counter: Counter,
    N_esp: int,
    N_gen: int,
    epsilon: float = 1e-9
) -> pd.DataFrame:
    """
    Calcula weirdness para un conjunto de candidatos (texto -> freq en especializado).
    freq_gen_counter es un Counter de n-gramas del corpus general (tuplas).
    
    Devuelve un DataFrame con:
    - term
    - freq_esp, rel_esp
    - freq_gen, rel_gen
    - weirdness
    """
    rows = []
    # Construimos un diccionario rápido para freq general en formato texto
    gen_text_freq = {ngram_to_text(k): v for k, v in freq_gen_counter.items()}
    
    for term, f_esp in cand_esp.items():
        f_gen = gen_text_freq.get(term, 0)
        r_esp = rel_freq(f_esp, N_esp)
        r_gen = rel_freq(f_gen, N_gen)
        w = r_esp / (r_gen + epsilon)
        rows.append((term, f_esp, r_esp, f_gen, r_gen, w))
    
    df = pd.DataFrame(rows, columns=["term","freq_esp","rel_esp","freq_gen","rel_gen","weirdness"])
    df = df.sort_values("weirdness", ascending=False).reset_index(drop=True)
    return df

df_uni = compute_weirdness(cand1_esp, freq1_gen, N_esp, N_gen)
df_bi  = compute_weirdness(cand2_esp, freq2_gen, N_esp, N_gen)
df_tri = compute_weirdness(cand3_esp, freq3_gen, N_esp, N_gen)

df_bi.head(10)
```

### 3.14. Miniglosario automático

Un “miniglosario” útil puede incluir:
- término candidato
- score (para justificar por qué aparece)
- ejemplos de concordancia (KWIC: *keyword in context*). ✅ KWIC ayuda a validar si el candidato “tiene sentido” como término.

KWIC muestra el término con palabras alrededor, por ejemplo:
- “… la tokenización es un paso …”
- “… métricas como BLEU …”

Esto ayuda a:
- entender significado por contexto
- verificar si realmente es un término


```python
def kwic(tokens: List[str], term_tokens: List[str], window: int = 5, max_examples: int = 5) -> List[str]:
    """
    Devuelve ejemplos KWIC para term_tokens dentro de una lista de tokens.
    - window: número de tokens a izquierda y derecha
    - max_examples: máximo de ejemplos a devolver
    
    Nota: buscamos coincidencia exacta de secuencia.
    """
    examples = []
    n = len(term_tokens)
    for i in range(len(tokens) - n + 1):
        if tokens[i:i+n] == term_tokens:
            left = " ".join(tokens[max(0, i-window):i])
            hit  = " ".join(tokens[i:i+n])
            right= " ".join(tokens[i+n:i+n+window])
            examples.append(f"{left}  >>{hit}<<  {right}".strip())
            if len(examples) >= max_examples:
                break
    return examples

# Probamos con un término
kwic(tokens_esp, ["modelo", "lenguaje"], window=4, max_examples=10)
```

**Visualización "bonita" del miniglosario**

En una clase, es habitual presentar:
- término
- score
- 2–3 ejemplos KWIC

La siguiente celda imprime una versión legible.


```python
def build_mini_glossary(df_ranked: pd.DataFrame, tokens_esp: List[str], top_n: int = 15) -> pd.DataFrame:
    rows = []
    for _, row in df_ranked.head(top_n).iterrows():
        term = row["term"]
        term_tokens = term.split()
        examples = kwic(tokens_esp, term_tokens, window=5, max_examples=3)
        rows.append({
            "term": term,
            "n": row["n"],
            "weirdness": row["weirdness"],
            "freq_esp": row["freq_esp"],
            "freq_gen": row["freq_gen"],
            "kwic_examples": examples
        })
    return pd.DataFrame(rows)

def print_glossary(glossary_df: pd.DataFrame):
    for i, row in glossary_df.iterrows():
        print(f"\n📌 #{i+1}  {row['term']}  (n={row['n']})")
        print(f"   🧮 weirdness={row['weirdness']:.2f} | esp={row['freq_esp']} | gen={row['freq_gen']}")
        if row["kwic_examples"]:
            print("   🔎 Ejemplos (KWIC):")
            for ex in row["kwic_examples"]:
                print("   -", ex)
        else:
            print("   ⚠️ Sin ejemplos KWIC (revisar filtros)")

glossary = build_mini_glossary(df_all_sorted, tokens_esp, top_n=15)
print_glossary(glossary)
```
