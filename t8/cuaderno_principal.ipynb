## 1️⃣ Pinceladas teóricas 🧠 

### 📌 ¿Qué es la clasificación temática?


Es asignar a un texto una etiqueta (clase) que describe su **tema o registro**:
- Jurídico (normas, contratos, resoluciones)
- Técnico (manuales, instrucciones, especificaciones)
- Divulgativo (explica temas al público general)
- Literario (relato, estilo artístico, metáforas)

### ✅ ¿Por qué “explicable”?

En PLN, muchos modelos clasifican muy bien pero no sabemos **por qué**.
Aquí buscamos lo contrario:
- pocas reglas
- indicadores claros
- justificación con palabras clave y recuentos visibles

> En Humanidades, esto es valioso: puedes discutir y argumentar el criterio.

### 🔑 Tres ideas base de la unidad

#### 1) Palabras clave
Ciertas palabras son “señales” de una temática.
- jurídico: *artículo, ley, disposición, sanción, parte, cláusula...*
- técnico: *instalación, dispositivo, configuración, voltaje, manual...*
- divulgativo: *explica, de manera sencilla, en general, por ejemplo...*
- literario: *susurró, horizonte, memoria, silencios, metáfora...*

#### 2) Vectores simples (TF)
Representamos cada texto como un **listado de frecuencias**:
- cuántas veces aparece cada palabra importante

#### 3) Similitud léxica (coseno, conceptual)
Dos textos son “parecidos” si:
- comparten muchas palabras importantes
- en proporciones similares

La similitud coseno se interpreta así:
- **1** → muy parecidos
- **0** → no se parecen (vocabulario distinto)

✅ No necesitas trigonometría: lo usamos como *medida de parecido*.

## 2️⃣ Enfoque práctico (pipeline) 👩‍💻

Vamos a hacer una clasificación temática explicable en 5 pasos:

1) ✂️ **Preprocesar**: tokenizar, minúsculas, quitar puntuación y stopwords  
2) 🧺 **Vectorizar**: TF (frecuencia de términos)  
3) 🧩 **Prototipos por clase**: crear un “texto prototipo” por temática (con keywords)  
4) 🧭 **Similitud coseno**: comparar el texto con cada prototipo  
5) 🏁 **Regla de decisión**: elegir la clase con mayor similitud + explicación (top evidencias)

📌 Salida:
- etiqueta predicha
- tabla de similitudes por clase
- evidencias: palabras que más contribuyen

## 4️⃣ Desarrollo práctico

### Preparación del entorno

```python
import re
from collections import Counter
from math import sqrt
from typing import List, Dict, Tuple

import pandas as pd
```

### Textos para clasificar

```python
texts_to_classify = [
    "De conformidad con el artículo 12 de la Ley, la parte demandada N°2 deberá cumplir la obligación en el plazo previsto.",
    "Para instalar el dispositivo, conecte el cable al puerto USB y configure el controlador siguiendo el manual de usuario.",
    "En este artículo explicamos de manera sencilla qué es la energía solar y por qué es importante para reducir emisiones.",
    "La noche cayó sobre el puerto, y las luces temblaron como luciérnagas en la memoria del viajero."
]

texts_to_classify
```

### 🔑 Perfiles por clase: palabras clave

```python
class_keywords = {
    "jurídico": [
        "artículo", "ley", "disposición", "cláusula", "parte", "demandada", "obligación",
        "procedimiento", "sanción", "conformidad", "plazo", "resolución", "contrato"
    ],
    "técnico": [
        "instalar", "dispositivo", "conecte", "puerto", "usb", "configurar", "controlador",
        "manual", "usuario", "sistema", "parámetro", "voltaje", "instrucciones"
    ],
    "divulgativo": [
        "explicamos", "manera", "sencilla", "por", "qué", "importante", "por ejemplo",
        "en general", "concepto", "básico", "introducción", "aprendemos", "comparar"
    ],
    "literario": [
        "noche", "memoria", "silencio", "susurro", "horizonte", "temblaron",
        "luciérnagas", "viajero", "sombra", "latido", "metáfora", "puerto"
    ],
}
```

### ✂️ Preprocesamiento con spaCy 🧠

```python
import spacy
from typing import List
from collections import Counter
from math import sqrt

nlp = spacy.load("es_core_news_sm")
```

```python
def preprocess(text: str) -> List[str]:
    """
    Preprocesamiento lingüístico con spaCy:
    - tokenización
    - lematización
    - eliminación de stopwords
    - solo palabras alfabéticas
    """
    doc = nlp(text.lower())
    tokens = [
        tok.lemma_
        for tok in doc
        if tok.is_alpha
        and not tok.is_stop
        and len(tok) >= 2
    ]
    return tokens

# Prueba rápida
preprocess(
    "De conformidad con el artículo 12 de la Ley, la parte demandada deberá cumplir."
)
```

### 🧺 Representación vectorial sencilla (TF)

```python
def tf_vector(tokens: List[str]) -> Counter:
    """
    Representa un texto como vector TF (frecuencia de lemas).
    """
    return Counter(tokens)

# Ejemplo
tf_vector(preprocess("Las leyes regulan la ley y la normativa vigente."))
```

### 🧭 Similitud coseno

```python
def cosine_similarity(v1: Counter, v2: Counter) -> float:
    """
    Similitud coseno entre dos vectores TF (Counters).
    """
    common = set(v1.keys()) & set(v2.keys())
    dot = sum(v1[t] * v2[t] for t in common)

    norm1 = sqrt(sum(v**2 for v in v1.values()))
    norm2 = sqrt(sum(v**2 for v in v2.values()))

    if norm1 == 0 or norm2 == 0:
        return 0.0
    return dot / (norm1 * norm2)
```

### 🧩 Prototipos temáticos (lemmatizados) 

```python
def build_class_prototypes(class_keywords: dict) -> dict:
    """
    Convierte las keywords de cada clase en prototipos TF lematizados.
    """
    prototypes = {}
    for label, keywords in class_keywords.items():
        tokens = []
        for kw in keywords:
            tokens.extend(preprocess(kw))
        prototypes[label] = tf_vector(tokens)
    return prototypes

prototypes = build_class_prototypes(class_keywords)
prototypes["jurídico"].most_common(10)
```

### 🏁 Regla de decisión explicable - clasificación de textos

```python
def top_evidence_terms(text_vec: Counter, proto_vec: Counter, top_k: int = 8):
    """
    Devuelve los lemas del texto que también están en el prototipo,
    ordenados por contribución simple.
    """
    contrib = []
    for t, f in text_vec.items():
        if t in proto_vec:
            contrib.append((t, f * proto_vec[t]))
    contrib.sort(key=lambda x: x[1], reverse=True)
    return contrib[:top_k]


def classify_text(text: str, prototypes: dict) -> dict:
    tokens = preprocess(text)
    vec = tf_vector(tokens)

    scores = {
        label: cosine_similarity(vec, proto)
        for label, proto in prototypes.items()
    }

    best_label = max(scores, key=scores.get)
    evidence = top_evidence_terms(vec, prototypes[best_label])

    return {
        "text": text,
        "tokens": tokens,
        "predicted_label": best_label,
        "scores": scores,
        "evidence": evidence
    }
```

```python
# Clasificar todos los textos
results = [classify_text(t, prototypes) for t in texts_to_classify]

# Construir tabla de similitudes
rows = []
for r in results:
    row = {
        "predicted_label": r["predicted_label"],
        "text": r["text"]
    }
    for label, score in r["scores"].items():
        row[f"sim_{label}"] = score
    rows.append(row)

# DataFrame final
df_scores = pd.DataFrame(rows)

df_scores
```

### 🧾 Informe de clasificación (spaCy) 

```python
def print_report(results):
    for i, r in enumerate(results, 1):
        print("\n" + "="*80)
        print(f"📄 TEXTO #{i}")
        print(r["text"])
        print(f"\n🏷️ Clase predicha: {r['predicted_label']}")
        print("\n📊 Similitudes:")
        for label, score in sorted(r["scores"].items(), key=lambda x: x[1], reverse=True):
            print(f" - {label:12s}: {score:.3f}")
        print("\n🔎 Evidencias (lemas compartidos):")
        if r["evidence"]:
            for term, weight in r["evidence"]:
                print(f" - {term} (peso={weight})")
        else:
            print(" - (sin coincidencias claras)")

print_report(results)
```
