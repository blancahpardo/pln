## 🧪 Actividad propuesta

### Instrucciones

**Clasificación temática explicable**

#### 🎯 Objetivo de la actividad
Aplicar el método de clasificación temática explicable desarrollado en el cuaderno para:
- clasificar textos reales,
- justificar las decisiones del sistema,
- reflexionar sobre las ventajas y límites del enfoque.

Esta actividad pone el foco no solo en el resultado, sino en la argumentación lingüística detrás de cada decisión.

---

#### 📝 Descripción de la tarea

**1️⃣ Preparación de los datos**

Cada estudiante (o grupo) deberá:

- Seleccionar **8–10 textos muy breves** —una o dos frases cada uno—, de forma que **entre todos** sumen 100–200 palabras (es decir, unas 10–25 palabras por texto, no 100–200 cada uno), repartidos entre al menos **3 temáticas** distintas, por ejemplo:
  - jurídico
  - técnico
  - divulgativo
  - literario
  - (u otras, si se justifican)

Los textos pueden proceder de:
- fragmentos de leyes o contratos,
- manuales o instrucciones técnicas,
- artículos divulgativos,
- textos narrativos o literarios breves.

📌 **Importante**: los textos deben ser auténticos (no inventados) y citar su procedencia.

---

**2️⃣ Ejecución del sistema**

Usando el cuaderno:

- Sustituye `texts_to_classify` por tu propio conjunto de textos.
- Ejecuta el pipeline completo:
  - preprocesamiento con spaCy,
  - vectorización TF,
  - similitud coseno,
  - clasificación por prototipos temáticos.
  - informe final

Obtén para cada texto:
- la **clase predicha**,
- las **similitudes con cada temática**,
- las **evidencias léxicas** (lemas compartidos).

---

**3️⃣ Análisis y justificación**

Para al menos 4 textos, describe brevemente:

- por qué el sistema los ha clasificado en esa temática,
- qué palabras o lemas han sido determinantes,
- si estás de acuerdo con la decisión o no.

👉 Aquí es clave usar el vocabulario del curso:
*frecuencia*, *palabras clave*, *similitud léxica*, *prototipo*, *evidencia*.

---

**📤 Entrega final al cliente**

La entrega deberá incluir:

1. Tabla de resultados, con:
   - texto (o identificador),
   - clase predicha,
   - similitudes por clase.
   Esta tabla es la correspondiente al paso en el que ejecutas la variable para generar `df_resultados`(como vimos en la parte práctica explicativa del cuaderno).

2. Comentario justificativo de al menos 4 textos (65-85 palabras cada uno).

3. Preguntas de reflexión (75-90 palabras cada una):
    - ¿En qué situaciones crees que este sistema explicable es preferible a un clasificador automático más complejo (por ejemplo, basado en redes neuronales), aunque sea menos preciso? Justifica tu respuesta con ejemplos del cuaderno.
    - ¿Cómo influyen las palabras clave elegidas en los resultados del sistema? ¿Qué ocurre si una temática está mal representada o si dos temáticas comparten vocabulario? Relaciónalo con el concepto de sesgo del modelo.
    - ¿Qué tipos de textos o situaciones crees que este método clasifica peor? Piensa, por ejemplo, en textos híbridos, irónicos, muy breves o muy especializados, y explica por qué el método basado en frecuencia y similitud puede fallar.
