## 🧪 Introducción al proceso de preprocesamiento de textos



El preprocesamiento de texto es una fase esencial en cualquier flujo de trabajo de procesamiento del lenguaje natural (PLN). Antes de aplicar modelos de traducción automática, análisis léxico o minería de datos lingüísticos, es imprescindible transformar los textos en bruto en representaciones limpias, estructuradas y homogéneas.

Esta unidad se centra en las principales técnicas de preprocesamiento: tokenización, normalización, eliminación de palabras vacías (stopwords) y lematización, con especial énfasis en su utilidad en contextos reales de traducción y localización. Aprenderás a construir pipelines de limpieza que te permitirán preparar corpus multilingües con criterios lingüísticos sólidos y con el nivel de calidad necesario para aplicaciones profesionales.

Este conocimiento es especialmente relevante para traductores especializados que colaboran en proyectos de tecnología lingüística o que desean intervenir en tareas de posedición, evaluación y entrenamiento de motores de traducción automática.

## 🧪 Introducción al análisis morfosintáctico



El análisis morfosintáctico es una etapa fundamental dentro del procesamiento del lenguaje natural (PLN). Consiste en asignar a cada palabra de un texto su categoría gramatical (o etiqueta PoS, Part-of-Speech) y analizar su función sintáctica dentro de la oración. Este nivel de análisis permite comprender la estructura del texto, facilitando tareas avanzadas como la extracción de información, traducción automática o generación de lenguaje natural.

En esta unidad, nos centraremos en aplicar técnicas de preprocesamiento lingüístico que son esenciales para preparar textos técnicos y científicos, especialmente en campos sensibles como el médico y farmacéutico. Utilizaremos **spaCy**, una biblioteca avanzada de Python, para realizar tokenización, etiquetado gramatical, eliminación de ruido y lematización.

La correcta preparación y análisis morfosintáctico del corpus mejora la calidad y precisión de modelos de traducción automática y otras aplicaciones de IA lingüística, garantizando que los sistemas interpreten correctamente la terminología especializada y eviten ambigüedades.

## 🔹 2. Carga y explotación del corpus

```python
!pip install --quiet spacy nltk

import nltk
nltk.download('stopwords')

!python -m spacy download es_core_news_sm

import spacy
from nltk.corpus import stopwords
import re

nlp = spacy.load("es_core_news_sm")

texto = """
La administración de ibuprofeno debe realizarse con precaución en pacientes con antecedentes de úlcera gástrica.
Los AINE pueden provocar efectos adversos, especialmente en pacientes de edad avanzada.
Se recomienda consultar con un profesional de la salud antes de iniciar cualquier tratamiento.
"""
print("Texto original:")
print(texto)
```

## 🔹 3. Preprocesamiento del corpus

```python
# Paso 1:
texto_normalizado = texto.lower()
print("Texto normalizado:")
print(texto_normalizado)
```

```python
# Paso 2:
texto_sin_puntuacion = re.sub(r'[^\w\s]', '', texto_normalizado)
print("Texto sin puntuación:", texto_sin_puntuacion)
```

```python
# Paso 3:
doc = nlp(texto_sin_puntuacion)
tokens = [token.text for token in doc]
print("Tokens:")
print(tokens)
```

```python
# Paso 4:

import nltk
nltk.download('stopwords')

from nltk.corpus import stopwords

stopwords_es = stopwords.words('spanish')
tokens_filtrados = [token for token in tokens if token not in stopwords_es]
print("Tokens sin stopwords:")
print(tokens_filtrados)
```

```python
# Paso 5:
lemmas = [token.lemma_ for token in doc if token.text not in stopwords_es]
print("Lemas (sin stopwords):")
print(lemmas)
```

```python
# Paso 6:
from collections import Counter

frecuencias = Counter(lemmas)

print("\nGlosario preliminar (orden alfabético):")
for palabra in sorted(frecuencias):
    print(f"- {palabra} ({frecuencias[palabra]})")
```

## 🔹 4. Análisis morfosintáctico y glosario técnico


```python
# Paso 4.1
doc = nlp(texto)

print("\nEtiquetas gramaticales (PoS):")
for token in doc:
    print(f"{token.text} - {token.pos_}")
```

```python
# Paso 4.2
from collections import Counter

sustantivos = [token.text.lower() for token in doc if token.pos_ == "NOUN"]

frecuencia_sustantivos = Counter(sustantivos)

print("\nSustantivos encontrados:")
for palabra, freq in frecuencia_sustantivos.most_common():
    print(f"{palabra}: {freq}")
```

```python
# Paso 4.3
categorias = [token.pos_ for token in doc]
conteo_categorias = Counter(categorias)
print("\nFrecuencia por categoría gramatical:")
for categoria, freq in conteo_categorias.items():
    if categoria in ["NOUN", "VERB", "ADJ"]:
        print(f"{categoria}: {freq}")
```

```python
# Paso 4.4
print("\nGlosario técnico (sustantivos + contexto):")
for sent in doc.sents:
    for palabra in frecuencia_sustantivos:
        if palabra in sent.text.lower():
            print(f"- [{palabra}] → {sent.text.strip()}")
```

```python
# Paso extra:
print("\nGlosario preliminar (orden alfabético):")
for palabra in sorted(frecuencia_sustantivos):
    print(f"- {palabra} ({frecuencia_sustantivos[palabra]})")
```
